# Comparing two chat models

This notebook compares `gpt-fixture-retired` with `gpt-fixture-4o` on a few prompts
and plots their scores.



```python
%pip install --quiet openai matplotlib
```

```python
import os

from openai import OpenAI

client = OpenAI()
# gpt-fixture-retired was the default until it retired; see the changelog.
MODEL = os.environ.get("MODEL", "gpt-fixture-retired")
CANDIDATE = "gpt-fixture-4o"
```

```python
response = client.responses.create(model=CANDIDATE, input="Say hello in French.")
print(response.model)
print(response.output_text)
```

```python
import matplotlib.pyplot as plt

# sunset-ignore-next-line reason="historical scores"
scores = {"gpt-fixture-retired": 0.71, "gpt-fixture-4o": 0.84}
plt.bar(list(scores), list(scores.values()))
plt.title("Accuracy by model")
```

## Results

`gpt-fixture-4o` scored higher, so it replaces gpt-fixture-retired from here on.
